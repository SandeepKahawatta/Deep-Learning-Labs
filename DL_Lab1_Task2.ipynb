### TASK 2

```python
import pandas as pd

try:
    df = pd.read_csv('/content/Au_nanoparticle_dataset.csv')
except FileNotFoundError:
    print("Error:'Au_nanoParticle_dataset.csv' is want to uploaded to your Colab session.")
```

1.Create a New Filtered DataFrame

```python

columns_to_keep = ['N_total', 'N_bulk', 'N_surface', 'R_avg']
new_df = df[columns_to_keep]

print("New DataFrame created with selected columns.")
```

2.Display the First 20 Samples

```python
print("First 20 samples of the new DataFrame:")
display(new_df.head(20))
```

3.Calculate Descriptive Statistics

```python
statistics = new_df.describe()

print("Mean, Standard Deviation, and Quartiles for each feature:")
display(statistics)
```

4.Plot Histograms in a 1x4 Layout

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 4, figsize=(20, 4), sharey=True)
fig.suptitle('Histograms of NanoParticle Features', fontsize=16)

for i, col_name in enumerate(new_df.columns):
    sns.histplot(data=new_df, x=col_name, ax=axes[i], bins=30)
    axes[i].set_title(col_name)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

5.Visualize with a Pairplot

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(new_df)
plt.show()
```

 6.Customize the Pairplot with PairGrid

```python
import seaborn as sns
import matplotlib.pyplot as plt

g = sns.PairGrid(new_df)

# a)
g.map_diag(sns.histplot, kde=True)

g.map_upper(sns.histplot)

# b)
g.map_lower(sns.kdeplot, cmap='Blues_d')

plt.suptitle('Customized PairGrid of NanoParticle Features', y=1.02)
plt.show()
```
